```python
import plotly.express as px
from plotly.subplots import make_subplots
import pandas as pd
```

```python
# load all_zf_metadata_combined.csv
zf_metadata = pd.read_csv('../all_zf_metadata_combined.csv',
                          parse_dates=['published_date']).dropna()
```

```python
# bulk	celseq	cytoseq	droncseq	dropseq	fluidigm	indrops	marsseq	matqseq	quartzseq	splitseq	superseq	microwellseq	scirnaseq	smarter	seqwell	strtseq	sortseq	slamseq	icell8	iclip	454	scartrace	10x	smartseq	generic-scrnaseq	generic-scrnaseq-only
colors = px.colors.qualitative.Vivid
colormap = {
    "unknown": "grey",
    "bulk": colors[0],
    "generic-scrnaseq-only": "black",
    "10x": colors[2],
    "smartseq": colors[3],
    "microwellseq": colors[4],
    "celseq": colors[5],
    "indrops": colors[6],
    "iclip": colors[7],
    "scirnaseq": colors[8],
    "slamseq": colors[9],
    "scartrace": colors[10],
    "fluidigm": "red",
    "sortseq": "red",
    "smarter": "grey", # not good
    "454": "yellow",
    "dropseq": "red", "quartzseq": "red",
    "cytoseq": "red", "splitseq":  "red", "superseq": "red", "seqwell": "red", "strtseq": "red", "marsseq": "red", "matqseq": "red", "icell8": "red", "droncseq": "red",
}
fig = px.scatter(zf_metadata, x='published_date', y='total_spots', color='technology',
                 color_discrete_map=colormap,
                 hover_data=['accession', 'technology'])
fig.update_traces(marker=dict(size=5))
fig.show()
```

```python
bins = 12 * 5
# stacked bar runs technology by year
hist = px.histogram(zf_metadata, x='published_date',
                   color='technology', color_discrete_map=colormap,
                   barmode='stack', nbins=bins,
                   hover_data=['published_date', 'technology'])

pie = px.pie(zf_metadata, names='technology',
             color='technology', color_discrete_map=colormap)
pie.update_traces(textposition='inside', textinfo='percent+label', showlegend=False)

fig = make_subplots(rows=1, cols=3, specs=[[{"type": "histogram", "colspan":2}, {},
                                           {"type": "pie", "colspan":1}]])
for d in hist.data:
    fig.add_trace(d, row=1, col=1)
fig.update_layout(barmode='stack', xaxis_title='Date', yaxis_title='SRA runs')
for d in pie.data:
    fig.add_trace(d, row=1, col=3)
fig.update_layout(title='Zebrafish SRA runs by technology described in metadata',
                  height=500, width=1200,
                  legend=dict(orientation='v', itemsizing='trace', tracegroupgap=0))
fig.write_image('zebrafish_sra_runs_by_technology.svg')
fig.show()
```

```python
bins = 12 * 5
# stacked bar reads technology by year
hist = px.histogram(zf_metadata, x='published_date', y='total_spots',
                   color='technology', color_discrete_map=colormap,
                   barmode='stack', nbins=bins,
                   hover_data=['published_date', 'technology'])

pie = px.pie(zf_metadata, values='total_spots', names='technology',
             color='technology', color_discrete_map=colormap)
pie.update_traces(textposition='inside', textinfo='percent+label', showlegend=False)

fig = make_subplots(rows=1, cols=3, specs=[[{"type": "histogram", "colspan":2}, {},
                                           {"type": "pie", "colspan":1}]])
for d in hist.data:
    fig.add_trace(d, row=1, col=1)
fig.update_layout(barmode='stack', xaxis_title='Date', yaxis_title='Total spots')
for d in pie.data:
    fig.add_trace(d, row=1, col=3)
fig.update_layout(title='Zebrafish spots (reads) in SRA by technology described in metadata',
                  height=500, width=1200,
                  legend=dict(orientation='v', itemsizing='trace', tracegroupgap=0))
fig.write_image('zebrafish_spots_by_technology.svg')
fig.show()
```
