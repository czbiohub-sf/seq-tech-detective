```python
import plotly.express as px
import pandas as pd
```

```python
# load all_zf_metadata_combined.csv
zf_metadata = pd.read_csv('../all_zf_metadata_combined.csv',
                          parse_dates=['published_date']).dropna()
```

```python
# bulk	celseq	cytoseq	droncseq	dropseq	fluidigm	indrops	marsseq	matqseq	quartzseq	splitseq	superseq	microwellseq	scirnaseq	smarter	seqwell	strtseq	sortseq	slamseq	icell8	iclip	454	scartrace	10x	smartseq	generic-scrnaseq	generic-scrnaseq-only
colors = px.colors.qualitative.Vivid
colormap = {
    "unknown": "grey",
    "bulk": colors[0],
    "generic-scrnaseq-only": "black",
    "10x": colors[2],
    "smartseq": colors[3],
    "microwellseq": colors[4],
    "celseq": colors[5],
    "indrops": colors[6],
    "iclip": colors[7],
    "scirnaseq": colors[8],
    "slamseq": colors[9],
    "scartrace": colors[10],
    "fluidigm": "red",
    "sortseq": "red",
    "smarter": "grey", # not good
    "454": "yellow",
    "dropseq": "red", "quartzseq": "red",
    "cytoseq": "red", "splitseq":  "red", "superseq": "red", "seqwell": "red", "strtseq": "red", "marsseq": "red", "matqseq": "red", "icell8": "red", "droncseq": "red",
}
```

```python
zf_metadata['r1'] = zf_metadata['read:length'].str.extract(r'0:(\d+)').astype(int)
zf_metadata['r2'] = zf_metadata['read:length'].str.extract(r'1:(\d+)').fillna(0).astype(int)
zf_metadata['r3'] = zf_metadata['read:length'].str.extract(r'2:(\d+)').fillna(0).astype(int)
zf_metadata['r4'] = zf_metadata['read:length'].str.extract(r'3:(\d+)').fillna(0).astype(int)

zf_metadata['has_r3r4'] =  (zf_metadata['r3'] + zf_metadata['r4']) > 0
zf_metadata['has_r3r4'] = zf_metadata['has_r3r4'].replace({True: 'has r3/r4', False: 'no r3/r4'})
```

```python

fig = px.scatter_3d(zf_metadata, x='published_date', y='r1', z='r2',
                    color='technology',
                    color_discrete_map=colormap, symbol='has_r3r4',
                    symbol_map={'has r3/r4': 'cross', 'no r3/r4': 'circle'},
                    size=[2]*70625,
                    hover_data=['accession', 'technology', 'r3', 'r4'],
                    range_y=[0, 170])
fig.update_layout(width=1500, height=1000)
fig.show()
```

```python
fig.write_html('read_length.html')
```
